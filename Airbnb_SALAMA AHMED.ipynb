```python
def get_listing_details(driver, wait):
    rating, reviews, is_fav = "N/A", "0", 0
    property_type = "N/A"
    amenities_text = "N/A"
    cleaning_fee = "N/A"

    # 1. Rating, Reviews, and Guest Favorite
    try:
        try:
            review_link = wait.until(EC.presence_of_element_located(
                (By.XPATH, "//a[contains(@href, 'reviews')] | //button[contains(., 'reviews')]")
            ))
        except:
            try:
                h1_parent = driver.find_element(By.TAG_NAME, "h1").find_element(By.XPATH, "./..").text
                if "New" in h1_parent:
                    rating = "New"
            except: pass
        else:
            link_text = review_link.get_attribute("textContent").strip()
            clean_link = " ".join(link_text.split())

            parent_text = review_link.find_element(By.XPATH, "./..").get_attribute("textContent").strip()
            clean_parent = " ".join(parent_text.split())

            if "Guest favorite" in clean_link or "Guest favorite" in clean_parent:
                is_fav = 1
                r_match = re.search(r"(\d\.\d{1,2})", clean_parent)
                if r_match: rating = r_match.group(1)
                rev_match = re.search(r"(\d+)\s*Review", clean_parent, re.IGNORECASE)
                if rev_match: reviews = rev_match.group(1)
            else:
                is_fav = 0
                if "New" in clean_parent:
                    rating = "New"
                else:
                    r_match = re.search(r"(\d\.\d{1,2})", clean_parent)
                    if r_match: rating = r_match.group(1)

                rev_match = re.search(r"(\d+)\s*review", clean_link, re.IGNORECASE)
                if not rev_match:
                    rev_match = re.search(r"(\d+)\s*review", clean_parent, re.IGNORECASE)
                if rev_match: reviews = rev_match.group(1)
    except: pass

    # 2. Property Type (Extra Feature)
    try:
        prop_elem = driver.find_element(By.XPATH, "//h2[contains(text(), 'Hosted by')]/preceding-sibling::div[1] | //div[@data-section-id='TITLE_DEFAULT']/following-sibling::div//div")
        property_type = prop_elem.text.split("·")[0].strip()
    except:
        try:
            prop_elem = driver.find_element(By.CSS_SELECTOR, "ol._70ke8g")
            property_type = prop_elem.text
        except:
            pass

    # 3. Amenities (Extra Feature)
    try:
        driver.execute_script("window.scrollBy(0, 1000);")
        time.sleep(0.5)
        amenities_section = driver.find_element(By.XPATH, "//div[@data-section-id='AMENITIES_DEFAULT']")
        amenities_text = amenities_section.text[:200] # Taking a snippet of available amenities
    except:
        pass

    # 4. Cleaning Fee (Extra Feature)
    try:
        fee_elem = driver.find_element(By.XPATH, "//span[contains(text(), 'Cleaning fee')]/following::span[1]")
        cleaning_fee = fee_elem.text
    except:
        pass

    return rating, reviews, is_fav, property_type, amenities_text, cleaning_fee

all_data = []
city_count = 0

for city in cities:
    city_count += 1

    URL = f"https://www.airbnb.com/s/{city}/homes?refinement_paths%5B%5D=%2Fhomes&date_picker_type=calendar&search_type=filter_change&flexible_trip_lengths%5B%5D=one_week&monthly_start_date=2026-02-01&monthly_length=3&monthly_end_date=2026-05-01&price_filter_input_type=2&channel=EXPLORE&price_filter_num_nights=5&min_bedrooms=2&selected_filter_order%5B%5D=min_bedrooms%3A2&selected_filter_order%5B%5D=room_types%3AEntire%20home%2Fapt&update_selected_filters=true&query={city}&search_mode=regular_search&room_types%5B%5D=Entire%20home%2Fapt"
    service = Service(ChromeDriverManager().install())
    driver = webdriver.Chrome(service=service, options=chrome_options)
    target_count = 50
    items_scraped_this_city = 0
    try:
        driver.get(URL)
        wait = WebDriverWait(driver, 8)
        while items_scraped_this_city < target_count:
            print(f"\n=== SCRAPING CITY {city_count}/{len(cities)}: {city} ===")
            try:
                wait.until(EC.presence_of_all_elements_located((By.CSS_SELECTOR, "div[data-testid='card-container']")))
            except: break
            listings = driver.find_elements(By.CSS_SELECTOR, "div[data-testid='card-container']")
            listing_urls = []
            for l in listings:
                try:
                    url = l.find_element(By.TAG_NAME, "a").get_attribute("href")
                    if url: listing_urls.append(url)
                except: continue
            main_window = driver.current_window_handle
            for url in listing_urls:
                if items_scraped_this_city >= target_count: break
                try:
                    driver.execute_script(f"window.open('{url}', '_blank');")
                    driver.switch_to.window(driver.window_handles[-1])
                    wait.until(EC.presence_of_element_located((By.TAG_NAME, "h1")))

                    # A. Name
                    try: name = driver.find_element(By.TAG_NAME, "h1").text
                    except: name = "No Title"

                    # B. Superhost
                    is_superhost = "No"
                    try:
                        driver.execute_script("window.scrollBy(0, 500);")
                        time.sleep(0.5)
                        body_text = driver.find_element(By.TAG_NAME, "body").text
                        if "Superhost" in body_text: is_superhost = "Yes"
                    except: pass

                    # C. Price
                    try:
                        price_element = driver.find_element(By.XPATH, "//span[contains(text(), '$') or contains(text(), '€') or contains(text(), 'per night')]")
                        price = price_element.text
                    except:
                        try:
                           price = driver.find_element(By.CSS_SELECTOR, "span._tyxjp1").text
                        except:
                           price = "N/A"

                    # D. Extended Details & Features (Rating, Reviews, Fav, Property Type, Amenities, Cleaning Fee)
                    rating, reviews, is_fav, property_type, amenities_text, cleaning_fee = get_listing_details(driver, wait)

                    all_data.append({
                        "Name": name,
                        "city": city,
                        "Price": price,
                        "Superhost": is_superhost,
                        "is_fav": is_fav,
                        "Rating": rating,
                        "Reviews": reviews,
                        "Property_Type": property_type,
                        "Amenities": amenities_text,
                        "Cleaning_Fee": cleaning_fee,
                        "URL": url
                    })
                    print(f"[{len(all_data)}] {price} | {city} | ★{rating} ({reviews}) | Fav: {is_fav} | SH: {is_superhost} | Type: {property_type}")
                    time.sleep(random.uniform(1.0, 2.0))
                    driver.close()
                    driver.switch_to.window(main_window)
                    items_scraped_this_city += 1

                except Exception:
                    if len(driver.window_handles) > 1:
                        driver.close()
                        driver.switch_to.window(main_window)
                    continue
            if items_scraped_this_city < target_count:
                try:
                    next_btn = driver.find_element(By.CSS_SELECTOR, "a[aria-label='Next']")
                    driver.execute_script("arguments[0].click();", next_btn)
                    time.sleep(4)
                except: break

    finally:
        print(f"\nFinished city: {city}. Total scraped so far: {len(all_data)}")
        driver.quit()

# --- SAVE TO CSV AT THE VERY END ---
df_final = pd.DataFrame(all_data)
df_final.to_csv("airbnb_new_scraped_features.csv", index=False)
print(f"\n SUCCESS! All data saved to 'airbnb_new_scraped_features.csv'. Total records: {len(df_final)}")
```

```python
import sqlite3
```

```python
conn = sqlite3.connect("airbnb_datawarehouse.db")
cursor = conn.cursor()
```

```python
Cities.to_sql("city",conn,if_exists="replace",index=False)
units.to_sql("units",conn,if_exists="replace",index=False)
main_data.to_sql("main_data",conn,if_exists="replace",index=False)
```

```python
cursor.execute("DROP TABLE IF EXISTS fact_listings;")
```

```python
cursor.execute("""
CREATE TABLE fact_listings AS
SELECT
    u.unit_id,
    c.city_id,
    m.Price,
    m.bedrooms,
    m.dist,
    m.Reviews,
    m.Rating,
    m.is_fav,
    m.Superhost
FROM main_data m
LEFT JOIN city c ON m.city = c.city_name
LEFT JOIN units u ON m.room_type = u.room_type AND m.person_capacity = u.person_capacity;
""")
conn.commit()
print("=== Data Warehouse & Fact Table Created Successfully via SQL! ===")
```

```python
query_test = """
SELECT
    c.city_name,
    COUNT(f.unit_id) AS total_listings,
    ROUND(AVG(f.Price), 2) as average_price
FROM fact_listings f
JOIN city c ON f.city_id = c.city_id
GROUP BY c.city_name
LIMIT 5;
"""
```

```python
airbnb_datawarehouse_test = pd.read_sql(query_test, conn)
print(airbnb_datawarehouse_test)
print(f"airbnb_datawarehouse_test shape: {airbnb_datawarehouse_test.shape}")
```

```python
tables_query = "SELECT name FROM sqlite_master WHERE type='table';"
tables_df = pd.read_sql(tables_query, conn)
print("=== Tables inside Data Warehouse ===")
print(tables_df)
```

```python
for table in tables_df['name']:
    count_query = f"SELECT COUNT(*) as total_rows FROM {table};"
    count_df = pd.read_sql(count_query, conn)
    print(f"Table '{table}' -> Total Rows: {count_df['total_rows'].values[0]}")
```

```python
validation_query = """
SELECT
    c.city_name,
    COUNT(f.unit_id) as properties_count,
    ROUND(AVG(f.Price), 2) as avg_price,
    ROUND(AVG(f.Rating), 2) as avg_rating
FROM fact_listings f
JOIN city c ON f.city_id = c.city_id
GROUP BY c.city_name;
"""
validation_df = pd.read_sql(validation_query, conn)
print("\n=== Data Warehouse Validation & Summary Table ===")
print(validation_df)
```
